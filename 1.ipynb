```python
def largest(A):
    n = len(A)
    m = len(A[0])
    h = [0] * m
    best = [0] * n

    for r in range(n):
        for c in range(m):
            if A[r][c] == '0':
                h[c] += 1
            else:
                h[c] = 0

        stack = []

        for c in range(m + 1):
            cur = h[c] if c < m else 0
            start = c

            while stack and stack[-1][1] > cur:
                s, height = stack.pop()
                best[r] = max(best[r], height * (c - s))
                start = s

            stack.append((start, cur))

        if r > 0:
            best[r] = max(best[r], best[r-1])

    return best
```

```python
def solution(A):

    def solve(A):
        n = len(A)
        m = len(A[0])

        h = [0] * m
        best = [0] * n

        for r in range(n):
            for c in range(m):
                if A[r][c] == '0':
                    h[c] += 1
                else:
                    h[c] = 0

            stack = []

            for c in range(m + 1):
                cur = h[c] if c < m else 0
                start = c

                while stack and stack[-1][1] > cur:
                    s, height = stack.pop()
                    area = height * (c - s)

                    best[r] = max(best[r], area)
                    start = s

                stack.append((start, cur))

            if r > 0:
                best[r] = max(best[r], best[r - 1])

        return best

    # Best rectangle in rows 0...r
    top = solve(A)

    # Best rectangle in rows r...N-1
    bottom = solve(A[::-1])
    bottom.reverse()

    ans = top[-1]

    # Split horizontally
    for r in range(len(A) - 1):
        ans = max(ans, top[r] + bottom[r + 1])

    # Transpose the matrix
    B = []

    for c in range(len(A[0])):
        row = ""
        for r in range(len(A)):
            row += A[r][c]
        B.append(row)

    # Best rectangle on left/right
    left = solve(B)

    right = solve(B[::-1])
    right.reverse()

    # Split vertically
    for c in range(len(B) - 1):
        ans = max(ans, left[c] + right[c + 1])

    return ans
```

```python

# faster
def solution(A):
    def f(A):
        m = len(A[0]); h = [0]*m; b = []
        for row in A:
            for j in range(m):
                h[j] = h[j]+1 if row[j]=='0' else 0
            s = []; mx = b[-1] if b else 0
            for j in range(m+1):
                x = h[j] if j<m else 0; st = j
                while s and s[-1][1] > x:
                    p,v = s.pop()
                    mx = max(mx, v*(j-p)); st = p
                s.append((st,x))
            b.append(mx)
        return b

    def check(A):
        x = f(A)
        y = f(A[::-1])[::-1]
        return max([x[-1]] + [x[i]+y[i+1] for i in range(len(A)-1)])

    ans = check(A)
    B = [''.join(x) for x in zip(*A)]
    return max(ans, check(B))
```
